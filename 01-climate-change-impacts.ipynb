# Climate Change Portfolio Post

Climate change is impacting the way people live around the world. Higher highs, lower lows, storms, and smoke – we’re all feeling the
effects of climate change. Here, we look at trends in temperature over time along the San Pedro River in Arizona.

### Climate Change Impacts:
How has climate change affected this area? 

>Before moving to Boulder, I lived in Northern New Mexico. Here are some ways that the region is being affected by climate change.
>* Climate models project that temperatures in Northern NM will increase by up to 4C by the 2050s. These increases in temperature and changes in precipitation type are projected to result in a 47% decrease in snow by 2050 (Bennett et al., 2020).
>* Increased piñon pine tree mortality as a result of climate-induced drought is resulting in decreased species richness, diversity, and abundance of avian communities in the region (Fair et al., 2018). 
>* Anthropogenic climate change caused significant increases in fuel aridity in forested areas throughout the western U.S. since the 1980s, resulting in large increases in the amount of land experiencing wildfires (Abatzoglou & Williams, 2016).
