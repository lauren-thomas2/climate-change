### Clean up time series plots by resampling

You may notice that your plot looks a little “fuzzy”. This happens when Python is trying to plot a value for every date, but the resolution of the image is too low to actually do that. You can address this issue by **resampling** the data, or summarizing it over a time period of your choice. In this case, we will resample annually, giving us one data point per year.

```python
### resample to mean annual temps
annual_temp_df = (

    ### resample df
    temp_units_df

    ### sample annually
    .resample('YS')

    ### calculate mean
    .mean()
)

annual_temp_df
```

### Plot the annual data

```python
annual_temp_df.plot(y = 'temp_c',
                  title = 'Mean Annual Temperature in Arizona',
                  xlabel = 'Year',
                  ylabel = 'Temperature ($^\circ$C)')
```

>* Mean annual temperatures are lower in the late 1800s/early 1900s than they are in the 2000s. This could be a result of climate change, but it is difficult to say that conclusively without further analysis/looking at the data more closely.
